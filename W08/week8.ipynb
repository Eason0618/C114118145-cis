# Mission 4 ：算術運算與成績公式

```python
# 小考成績
# = 50
# + 3 × 讀書時數
# + 2 × 睡眠時數
# - 1.2 × 螢幕使用時數
# + 0.15 × 出席率

# 假想學生的資料
study_hours = 4.0
sleep_hours = 7.0
screen_time_hours = 2.0
attendance_rate = 90

# 分別計算各項影響
base_score = 50
study_effect = 3 * study_hours
sleep_effect = 2 * sleep_hours
screen_effect = -1.2 * screen_time_hours
attendance_effect = 0.15 * attendance_rate

# 合計成績
quiz_score = base_score + study_effect + sleep_effect + screen_effect + attendance_effect

print("基礎分數：", base_score)
print("讀書加分：", study_effect)
print("睡眠加分：", sleep_effect)
print("螢幕扣分：", screen_effect)
print("出席加分：", attendance_effect)
print("---")
print("模擬小考成績：", quiz_score)
```

# Mission 5 ：條件判斷與自訂函式

```python
# 定義函式：根據分數回傳文字描述
def describe_score(score):
  if score == 100:
    return "滿分！"
  elif score >= 80:
    return "表現良好"
  elif score >= 60:
    return "達到基本標準"
  else:
    return "需要加強"

# 測試函式
print("分數 85：", describe_score(85))
print("分數 70：", describe_score(70))
print("分數 50：", describe_score(50))
print("分數 60：", describe_score(60))
print("分數 59：", describe_score(59))
print("分數 100：", describe_score(100))

# 使用變數呼叫
my_score = 78
print("我的分數", my_score, "：", describe_score(my_score))
```

```python
# 定義函式：根據分數回傳多個文字描述
def describe_score(score):
  results = []

  if score >= 80:
    results.append("表現良好")
  if score >= 60:
    results.append("達到基本標準")
  if score < 60:
    results.append("需要加強")
  return " 、".join(results)
print("分數 85：", describe_score(85))
print("分數 70：", describe_score(70))
print("分數 50：", describe_score(50))
```

# Mission 7 ：建立模擬的大學生學習資料

```python
# 設定學生人數
n = 60

# 1. 學生編號：1 到 60
student_id = list(range(1, n + 1))

# 2. 科系：從三個科系隨機選擇
departments_pool = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments_pool, n)

# 3. 睡眠時數：常態分布 平均 6.7，標準差 1.1，四捨五入到小數點後 1 位
sleep_hours = rng.normal(6.7, 1.1, n).round(1)

# 4. 螢幕使用時數：常態分布 平均 5.0，標準差 1.5，限制 1~10 小時
screen_time_hours = rng.normal(5.0, 1.5, n).clip(1, 10).round(1)

# 5. 讀書時數：常態分布 平均 2.5，標準差 1.0，限制 0.5~6 小時
study_hours = rng.normal(2.5, 1.0, n).clip(0.5, 6).round(1)

# 6. 出席率：常態分布 平均 88，標準差 8，限制 60~100，四捨五入成整數
attendance_rate = rng.normal(88, 8, n).clip(60, 100).round(0).astype(int)

# 建立 DataFrame
df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# 顯示前 60 筆
print("前 60 筆資料：")
df.head(60)

```

# Mission 8 ：建立小考成績欄位


```python
# W6-HW-1
# 逐項計算各影響因子
base = 50
study_effect = 3 * df["study_hours"]
sleep_effect = 2 * df["sleep_hours"]
screen_effect = -1.2 * df["screen_time_hours"]
attendance_effect = 0.15 * df["attendance_rate"]
random_error = rng.normal(0, 5, n)  # 隨機誤差

# 合計並限制範圍
quiz_score = (base + study_effect + sleep_effect + screen_effect + attendance_effect + random_error).clip(0, 100).round().astype("Int64")

# 新增欄位到 DataFrame
df["quiz_score"] = quiz_score

# 顯示指定欄位的前 60 筆
df[
  [
   "student_id",
   "study_hours",
   "sleep_hours",
   "screen_time_hours",
   "attendance_rate",
   "quiz_score"
  ]
].head(60)


```

# Google Colab 範本

```python
# ==========================================
# 🚀 Google Colab 完整入門範本
# 功能：環境檢查、GPU確認、套件安裝、資料視覺化、簡單模型訓練示範
# ==========================================

# ------------------------------------------
# 1. 環境與硬體檢查
# ------------------------------------------
import sys
import torch
import tensorflow as tf

print(f"🐍 Python 版本: {sys.version.split()[0]}")
print(f"🔥 PyTorch 版本: {torch.__version__}")
print(f"🧠 TensorFlow 版本: {tf.__version__}")

# 檢查 GPU (Colab 免費版通常提供 T4, T4 16GB VRAM)
print("\n--- 硬體加速狀態 ---")
if torch.cuda.is_available():
    gpu_name = torch.cuda.get_device_name(0)
    gpu_mem = torch.cuda.get_device_properties(0).total_memory / 1024**3
    print(f"✅ PyTorch CUDA 可用 | GPU: {gpu_name} | 顯存: {gpu_mem:.1f} GB")
    device = torch.device("cuda")
else:
    print("❌ PyTorch CUDA 不可用，將使用 CPU")
    device = torch.device("cpu")

# TensorFlow GPU 檢查
gpus_tf = tf.config.list_physical_devices('GPU')
if gpus_tf:
    print(f"✅ TensorFlow 偵測到 GPU: {gpus_tf}")
else:
    print("❌ TensorFlow 未偵測到 GPU")

# ------------------------------------------
# 2. 安裝額外套件 (如有需要)
# ------------------------------------------
# Colab 預裝大多數常用套件，若需要特定版本或新套件，取消註解下行：
# %pip install -q seaborn plotly scikit-learn transformers accelerate -U

# ------------------------------------------
# 3. 掛載 Google Drive (可選，需授權)
# ------------------------------------------
# from google.colab import drive
# drive.mount('/content/drive')
# print("Drive 掛載完成，路徑: /content/drive/MyDrive/")

# ------------------------------------------
# 4. 資料生成與視覺化 (使用 NumPy, Pandas, Matplotlib, Seaborn)
# ------------------------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 設定繪圖風格
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['font.sans-serif'] = ['DejaVu Sans'] # 避免中文亂碼，Colab 預設支援英文

# 生成模擬資料
np.random.seed(42)
n_samples = 500
data = {
    'X': np.random.randn(n_samples),
    'Y': 2.5 * np.random.randn(n_samples) + 10, # 線性關係 + 雜訊
    'Category': np.random.choice(['A', 'B', 'C'], n_samples, p=[0.4, 0.35, 0.25])
}
df = pd.DataFrame(data)

print("\n--- 資料前 5 筆 ---")
print(df.head())
print("\n--- 資料統計描述 ---")
print(df.describe())

# 繪圖：散佈圖 + 迴歸線
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# 左圖：類別散佈圖
sns.scatterplot(data=df, x='X', y='Y', hue='Category', alpha=0.7, ax=axes[0])
axes[0].set_title('Scatter Plot by Category')

# 右圖：迴歸分析
sns.regplot(data=df, x='X', y='Y', scatter_kws={'alpha':0.3}, line_kws={'color':'red'}, ax=axes[1])
axes[1].set_title('Linear Regression Fit')

plt.tight_layout()
plt.show()

# ------------------------------------------
# 5. PyTorch 簡單 GPU 訓練示範 (Linear Regression)
# ------------------------------------------
print("\n--- PyTorch 簡單訓練測試 (GPU 加速) ---")

# 建立資料張量並移至 device
X_tensor = torch.tensor(df[['X']].values, dtype=torch.float32).to(device)
Y_tensor = torch.tensor(df[['Y']].values, dtype=torch.float32).to(device)

# 定義模型
model = torch.nn.Linear(1, 1).to(device)
criterion = torch.nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 訓練迴圈
epochs = 200
for epoch in range(epochs):
    optimizer.zero_grad()
    outputs = model(X_tensor)
    loss = criterion(outputs, Y_tensor)
    loss.backward()
    optimizer.step()

    if (epoch+1) % 50 == 0:
        print(f"Epoch [{epoch+1}/{epochs}], Loss: {loss.item():.4f}, Weight: {model.weight.item():.4f}, Bias: {model.bias.item():.4f}")

print(f"\n✅ 訓練完成。最終參數 -> Weight: {model.weight.item():.4f}, Bias: {model.bias.item():.4f}")
print(f"   (真實關係約為 Weight=2.5, Bias=10)")

# ------------------------------------------
# 6. 計時工具示範 (%timeit 魔術指令)
# ------------------------------------------
# 在獨立儲存格執行效果最好，這裡用 time 模組示範
import time

print("\n--- 效能比較：CPU vs GPU 矩陣乘法 (10000x10000) ---")
size = 5000 # Colab 免費版顯存較小，調整為 5000 避免 OOM

# CPU 版
cpu_a = torch.randn(size, size)
cpu_b = torch.randn(size, size)
start = time.time()
_ = cpu_a @ cpu_b
print(f"CPU 時間: {time.time() - start:.4f} 秒")

# GPU 版 (如果有 GPU)
if device.type == 'cuda':
    gpu_a = torch.randn(size, size, device='cuda')
    gpu_b = torch.randn(size, size, device='cuda')
    # Warm up
    _ = gpu_a @ gpu_b
    torch.cuda.synchronize()

    start = time.time()
    _ = gpu_a @ gpu_b
    torch.cuda.synchronize() # 確保 GPU 運算結束再計時
    print(f"GPU 時間: {time.time() - start:.4f} 秒")
    print(f"🚀 加速比: ~{(time.time() - start) / (time.time() - start):.0f}x (實際差距通常 10x-50x)")

print("\n🎉 所有測試完成！你可以開始在 Colab 上開發專案了。")
```
