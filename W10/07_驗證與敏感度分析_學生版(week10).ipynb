## 共用資料準備
以下在記憶體重建與既有教材相同的資料。執行後可以直接開始本單元；不會覆寫 CSV。


````python
# 匯入套件：numpy 建立隨機資料，pandas 管理表格，pyplot 繪圖
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 固定種子，並維持呼叫順序，以重現原教材的 60 筆模擬資料
rng = np.random.default_rng(42)
n = 60
departments = ["資訊管理", "企業管理", "金融管理"]
# 每個字典鍵是一個欄位；對應的值是 60 名學生的資料
data = {
    "student_id": range(1, n + 1),
    "department": rng.choice(departments, n),
    "sleep_hours": rng.normal(6.7, 1.1, n).round(1),
    "screen_time_hours": rng.normal(5.0, 1.5, n).clip(1, 10).round(1),
    "study_hours": rng.normal(2.5, 1.0, n).clip(0.5, 6).round(1),
    "attendance_rate": rng.normal(88, 8, n).clip(60, 100).round(0),
}
df = pd.DataFrame(data)
# 成績依設定公式與隨機誤差建立；關係來自模擬規則，不能視為因果發現
noise = rng.normal(0, 5, n)
score = (50 + 3 * df["study_hours"] + 2 * df["sleep_hours"]
         - 1.2 * df["screen_time_hours"] + 0.15 * df["attendance_rate"] + noise)
df["quiz_score"] = score.clip(0, 100).round(0)
# 故意在計分後加入資料品質問題；無法從表格還原計分時原始值
df.loc[[5, 17], "sleep_hours"] = np.nan
df.loc[8, "screen_time_hours"] = 15.0
print("模擬資料形狀（列, 欄）：", df.shape)
````

````python
#例子
# 方案 A：全樣本（保留 15 小時）
import statsmodels.api as sm
X_all = sm.add_constant(all_pairs["screen_time_hours"])
y_all = all_pairs["quiz_score"]
model_all = sm.OLS(y_all, X_all).fit()

# 方案 B：排除 >10
X_sub = sm.add_constant(within_range["screen_time_hours"])
y_sub = within_range["quiz_score"]
model_sub = sm.OLS(y_sub, X_sub).fit()

# 比較關鍵指標
print("=== 方案 A（全樣本）===")
print(model_all.summary().tables[1])  # 係數表
print(f"R² = {model_all.rsquared:.4f}")

print("\n=== 方案 B（排除 >10）===")
print(model_sub.summary().tables[1])
print(f"R² = {model_sub.rsquared:.4f}")

# 觀察：screen_time_hours 係數變化多少？p-value 是否翻轉？R² 差多少？
````

敏感度分析的核心目的不是「找出正確答案」，而是「量化不確定性」。

在本單元中，第 9 位學生的 screen_time_hours 在計分完成後才被修改成 15.0（原始值約 7.3 已不可考）。這導致表格呈現「事後的錯誤輸入」與「事前的正確分數」同列共存。當我們比較「全保留（方案 A）」與「移除極端值（方案 B）」時，相關係數從 -0.327 變為 -0.262（相對變化 ~19.7%），雖然負相關方向未變，但效果大小明顯受單一極端值影響。敏感度分析告訴我們：「結論會隨處理方式擺動」，卻無法告訴我們「哪個版本是真理」，因為真相（原始值）已隨資料生成順序消失在歷史中。
